## La anomalía de modificación

```python
# El encuestador E04 cambia de telefono. Actualizamos... y una fila se nos escapa.
copia = df.copy()
filas_de_e04 = copia.index[copia["encuestador_legajo"] == "E04"]
print(f"filas que hay que cambiar: {len(filas_de_e04)}")

# Cambiamos todas menos la ultima. Un filtro mal puesto, un reemplazo que no llego al final.
copia.loc[filas_de_e04[:-1], "encuestador_telefono"] = "3407-500100"

copia[copia["encuestador_legajo"] == "E04"]["encuestador_telefono"].value_counts()
```

Dieciséis filas con el teléfono nuevo, una con el viejo.

Ahora el archivo dice dos teléfonos para la misma persona, y **los dos parecen igual de
legítimos**. No hay ningún error visible: no hay una celda roja, no hay un mensaje, no falta nada.
Solo hay dos verdades donde había una.

Ese es el costo real de la redundancia.

## La anomalía de inserción

```python
# El municipio incorpora Pergamino, que esta en Buenos Aires. Todavia no se
# visito ningun hogar. ¿Donde se anota?
pergamino = {"id_hogar": None, "localidad": "Pergamino", "provincia": "Buenos Aires",
             "fecha_visita": None, "encuestador_legajo": None, "encuestador_nombre": None,
             "encuestador_telefono": None, "personas": None, "ingreso": None,
             "cobertura_salud": None, "satisfaccion": None, "servicios": None}

# Concatenar una fila casi toda vacia emite un FutureWarning en pandas 2.x, y
# no queremos ruido en clase. El aviso es de dtypes, no del dato.
con_pergamino = df.copy()
with warnings.catch_warnings():
    warnings.simplefilter("ignore", FutureWarning)
    con_pergamino.loc[len(con_pergamino)] = pergamino
con_pergamino.tail(2)
```

Para registrar que Pergamino existe hubo que inventar una fila de visita con casi todo vacío: solo `localidad` y `provincia` tienen algo, y eso es justamente el punto.

Esa fila falsa ahora está en el archivo, y va a aparecer en cada conteo que alguien haga:

```python
print("visitas, contando la fila inventada:", len(con_pergamino))
print("visitas de verdad:                    ", len(df))
print()
print("promedio de personas con la fila falsa:", round(con_pergamino["personas"].mean(), 2))
print("promedio de personas sin ella:         ", round(df["personas"].mean(), 2))
```

El promedio no cambió, porque pandas saltea los vacíos al promediar. El conteo de visitas sí:
dice 46 donde hay 45.

**Cuál de los dos se equivoca depende de qué función use quien consulte**, y eso es peor que
equivocarse siempre: nadie puede predecir el error. El archivo no puede representar una localidad
sin hogares, y cualquier intento de forzarlo deja basura adentro.

# 5. La búsqueda de la clave

Apunte, sección 6.5. Los cuatro pasos, en orden.

```python
# Paso 1: ¿alguna columna sola tiene tantos valores distintos como filas hay?
resumen = pd.DataFrame({
    "valores_distintos": df.nunique(),
    "filas": len(df),
})
resumen["¿sirve sola?"] = resumen["valores_distintos"] == resumen["filas"]
resumen
```

Ninguna. `id_hogar` tiene 37 valores para 45 filas, porque ocho hogares aparecen dos veces.
`fecha_visita` tiene 28, porque varios días hubo más de una visita.

Ninguna columna sola identifica una fila.

```python
# Paso 2: probamos pares. Cuantos valores distintos tiene cada combinacion de dos columnas.
from itertools import combinations

candidatos = []
for a, b in combinations(df.columns, 2):
    distintos = df[[a, b]].drop_duplicates().shape[0]
    if distintos == len(df):
        candidatos.append((a, b, distintos))

pd.DataFrame(candidatos, columns=["columna A", "columna B", "combinaciones distintas"])
```

Salieron catorce pares, y trece son casualidad.

El motivo es que dos columnas tienen casi un valor por fila: `ingreso` tiene 42 valores distintos
en 45 filas, y `fecha_visita` tiene 28. Cualquier columna emparejada con una de esas dos alcanza a
distinguir las 45 filas, y no porque identifique nada: porque el ingreso de un hogar es casi
único.

El único par que sirve es **`id_hogar` más `fecha_visita`**, y se reconoce por el paso 4: es el
único donde las dos columnas identifican **por regla** y no por coincidencia. Ningún operativo
garantiza que dos hogares no compartan el ingreso.

Esa es la clave, y es **compuesta**: hacen falta las dos columnas.

```python
# Paso 3: ¿es minima? Si el par ya identifica, agregar una tercera no aporta nada.
par = df[["id_hogar", "fecha_visita"]].drop_duplicates().shape[0]
trio = df[["id_hogar", "fecha_visita", "personas"]].drop_duplicates().shape[0]

print(f"el par identifica:  {par} de {len(df)} filas")
print(f"el trío identifica: {trio} de {len(df)} filas")
print()
print("El trío no mejora nada, así que la clave es el par y no el trío.")
```

**Paso 4: la pregunta que los datos no contestan.**

¿Puede haber dos visitas al mismo hogar el mismo día? Si el operativo lo permite, esta clave se
rompe con la fila 46, y hace falta una clave sustituta `id_visita`.

Eso no se decide mirando la tabla. Se decide preguntándole a quien organizó el relevamiento. **Los
datos que tenés muestran lo que pasó, no lo que puede pasar.**

```python
# Lo que la clave revela: que una fila es una visita y no un hogar.
repetidos = df["id_hogar"].value_counts()
print("hogares con una sola visita:", (repetidos == 1).sum())
print("hogares con dos visitas:    ", (repetidos == 2).sum())
print()
df[df["id_hogar"] == 1][["id_hogar", "fecha_visita", "personas", "ingreso", "localidad"]]
```

Las dos visitas del hogar 1. `personas` cambió y `localidad` no.

Esa asimetría es la que va a resolver la clase 3: hay columnas que dependen de las dos partes de la
clave y columnas que dependen de una sola.

# 6. Las dos tablas, y el cruce

Apunte, sección 3.4.

```python
# Separamos localidad en su propia tabla, con una clave sustituta.
# El orden es el de aparicion en los datos: Ramallo, Cordoba, Concepcion.
localidad = (df[["localidad", "provincia"]]
             .drop_duplicates()
             .reset_index(drop=True))
localidad.insert(0, "id", range(1, len(localidad) + 1))
localidad = localidad.rename(columns={"localidad": "nombre"})
localidad
```

Tres filas. El hecho "Ramallo está en Buenos Aires" ahora está escrito **una vez**.

```python
# Y la tabla de hogares, que guarda solo una referencia.
hogar = (df[["id_hogar", "localidad"]]
         .drop_duplicates()
         .merge(localidad, left_on="localidad", right_on="nombre")
         [["id_hogar", "id"]]
         .rename(columns={"id_hogar": "id", "id": "localidad_id"})
         .sort_values("id")
         .reset_index(drop=True))

print(f"{len(hogar)} hogares")
hogar.head()
```

37 hogares, cada uno con un número que apunta a su localidad.

Ahora las tres anomalías. Cambiar la provincia de Ramallo es cambiar **una** celda. Agregar
Pergamino es agregar **una** fila a `localidad`, sin inventar ninguna visita. Y borrar las visitas a
Concepción no borra Concepción.

```python
# Pergamino, ahora si.
nueva = pd.DataFrame([{"id": 4, "nombre": "Pergamino", "provincia": "Buenos Aires"}])
pd.concat([localidad, nueva], ignore_index=True)
```

Una fila, sin nada vacío, sin inventar una visita que no existió. La anomalía de inserción
desapareció.

## El cruce: comprobar que no se perdió nada

```python
# Volvemos a juntar las dos tablas y comparamos contra el original.
reconstruido = (hogar
                .merge(localidad, left_on="localidad_id", right_on="id", suffixes=("", "_loc"))
                [["id", "nombre", "provincia"]]
                .rename(columns={"id": "id_hogar", "nombre": "localidad"})
                .sort_values("id_hogar")
                .reset_index(drop=True))

original = (df[["id_hogar", "localidad", "provincia"]]
            .drop_duplicates()
            .sort_values("id_hogar")
            .reset_index(drop=True))

print("¿Es idéntico al original?", reconstruido.equals(original))
reconstruido.head()
```

`True`. El cruce devolvió exactamente los 37 hogares con su localidad y su provincia.

Esa es la prueba de que la separación estuvo bien hecha: **si al volver a juntar no vuelve lo mismo,
la separación rompió el dato en vez de organizarlo.** La clase 2 lo usa como regla y la clase 3 lo
usa como verificación final.

La función `merge` de pandas es la que hace el cruce. Es la única función de pandas que este taller
necesita explicar, y en una base de datos se llama *join*.

# Cierre

Seis bloques, y una conclusión:

| Lo que vimos | Dónde queda |
|---|---|
| Un hecho escrito veinte veces | la redundancia |
| Un cambio que llegó a 16 de 17 filas | la anomalía de modificación |
| Pergamino sin dónde ser anotado | la anomalía de inserción |
| Tucumán borrado sin que nadie lo pidiera | la anomalía de borrado |
| `personas = -3` que entró sin una queja | el dominio que nadie controla |
| `id_hogar` más `fecha_visita` | la clave compuesta |
| El cruce que devolvió lo mismo | la prueba de que separar estuvo bien |

Ninguno de esos problemas se arregla limpiando el archivo. El archivo está limpio: lo dejó así el
T8001. Se arreglan cambiando dónde vive cada dato.

**La clase 2 es el método para decidir dónde va cada uno.**
