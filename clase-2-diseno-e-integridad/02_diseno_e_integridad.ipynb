# 2. Atributo o entidad

Apunte, sección 3.3. Las tres preguntas, aplicadas a dos columnas.

```python
def examinar(columna, tiene_propiedades_propias):
    valores = df[columna].dropna().unique()
    print(f"columna: {columna}")
    print(f"  valores distintos: {len(valores)} -> {sorted(valores)[:5]}")
    print(f"  ¿tiene propiedades propias?  {tiene_propiedades_propias}")

examinar("localidad", "SÍ: la provincia")
print()
examinar("cobertura_salud", "no")
```

`localidad` tiene una propiedad propia, la provincia. `cobertura_salud` guarda `si` o `no` y no
tiene nada más.

Las otras dos preguntas del apunte:

| | `localidad` | `cobertura_salud` |
|---|---|---|
| ¿Tiene propiedades propias? | sí, la provincia | no |
| ¿Puede existir sin filas que la referencien? | sí, Pergamino sin ningún hogar todavía | no |
| ¿Hay que administrar la lista de valores válidos? | sí, se agregan localidades a medida que el operativo crece | no, son dos palabras que no se mueven nunca |

`localidad` es **entidad**. `cobertura_salud` es **atributo**, y le alcanza un dominio.

> La regla corta del apunte: si lo único que necesitás es limitar los valores posibles, alcanza un
> dominio. Si además tiene datos propios o tiene que poder existir sola, es una entidad.

# 4. Las seis tablas

Apunte, secciones 4.1 a 4.4. Aplicamos las cuatro reglas de pasaje.

### Regla 1: cada entidad es una tabla

```python
# El orden es el de aparicion en los datos: Ramallo, Cordoba, Concepcion.
localidad = (df[["localidad", "provincia"]].drop_duplicates()
             .reset_index(drop=True)
             .rename(columns={"localidad": "nombre"}))
localidad.insert(0, "id", range(1, len(localidad) + 1))

encuestador = (df[["encuestador_legajo", "encuestador_nombre", "encuestador_telefono"]]
               .drop_duplicates().sort_values("encuestador_legajo").reset_index(drop=True)
               .rename(columns={"encuestador_legajo": "legajo",
                                "encuestador_nombre": "nombre",
                                "encuestador_telefono": "telefono"}))

servicio = pd.DataFrame({"id": [1, 2, 3, 4],
                         "nombre": ["agua", "luz", "gas", "cloacas"]})

print("localidad");   print(localidad.to_string(index=False)); print()
print("encuestador"); print(encuestador.to_string(index=False)); print()
print("servicio");    print(servicio.to_string(index=False))
```

### Regla 2: el vínculo uno a muchos pone la clave foránea en el lado «muchos»

```python
# hogar: recibe localidad_id
hogar = (df[["id_hogar", "localidad"]].drop_duplicates()
         .merge(localidad, left_on="localidad", right_on="nombre")
         [["id_hogar", "id"]]
         .rename(columns={"id_hogar": "id", "id": "localidad_id"})
         .sort_values("id").reset_index(drop=True))

# visita: recibe hogar_id y encuestador_legajo. Su clave es el par hogar_id + fecha.
visita = (df[["id_hogar", "fecha_visita", "encuestador_legajo", "personas",
              "ingreso", "cobertura_salud", "satisfaccion"]]
          .rename(columns={"id_hogar": "hogar_id", "fecha_visita": "fecha"})
          .sort_values(["hogar_id", "fecha"]).reset_index(drop=True))

print(f"hogar:  {len(hogar)} filas")
print(f"visita: {len(visita)} filas")
print()
visita.head(4)
```

Fijate qué **no** tiene la tabla `visita`: no tiene `localidad`, no tiene `provincia`, no tiene
el nombre ni el teléfono del encuestador.

La localidad no se fue por comodidad. Se fue porque **nunca fue un dato de la visita**: es del
hogar.

### Regla 3: el vínculo muchos a muchos crea una tabla nueva

```python
# La columna servicios guarda hasta cuatro valores en una celda. Cada valor es una fila.
pares = (df[["id_hogar", "servicios"]].drop_duplicates()
         .assign(servicios=lambda d: d["servicios"].str.split(";"))
         .explode("servicios")
         .rename(columns={"id_hogar": "hogar_id", "servicios": "nombre"}))

hogar_servicio = (pares.merge(servicio, on="nombre")
                  [["hogar_id", "id"]]
                  .rename(columns={"id": "servicio_id"})
                  .drop_duplicates()
                  .sort_values(["hogar_id", "servicio_id"]).reset_index(drop=True))

print(f"hogar_servicio: {len(hogar_servicio)} filas")
hogar_servicio.head(6)
```

88 filas, una por cada par de hogar y servicio. El hogar 1 tiene agua, luz y gas: tres filas.

Su clave primaria es el **par** de las dos columnas, y las dos son clave foránea.

```python
# El esquema completo, en numeros.
esquema = pd.DataFrame([
    ("localidad",      len(localidad),      len(localidad.columns)),
    ("encuestador",    len(encuestador),    len(encuestador.columns)),
    ("hogar",          len(hogar),          len(hogar.columns)),
    ("visita",         len(visita),         len(visita.columns)),
    ("servicio",       len(servicio),       len(servicio.columns)),
    ("hogar_servicio", len(hogar_servicio), len(hogar_servicio.columns)),
], columns=["tabla", "filas", "columnas"])
esquema["celdas"] = esquema["filas"] * esquema["columnas"]

print(esquema.to_string(index=False))
print()
print(f"tabla plana:  {len(df):3} filas x {len(df.columns)} columnas = {len(df)*len(df.columns)} celdas")
print(f"seis tablas:  {esquema['filas'].sum():3} filas               = {esquema['celdas'].sum()} celdas")
```

**594 celdas contra 540.** El diseño normalizado ocupa más, no menos.

Normalizar no es comprimir. Lo que se gana es otra cosa:

```python
for hecho, col in [("provincia de una localidad", "provincia"),
                   ("teléfono de un encuestador", "encuestador_telefono")]:
    en_plana = len(df)
    unicos = df[col].nunique()
    print(f"{hecho}:")
    print(f"   en la tabla plana: {en_plana} celdas")
    print(f"   en el diseño:      {unicos} celdas")
```

De 45 celdas a 3, y de 45 a 4.

Visto por hecho: que Ramallo está en Buenos Aires estaba escrito veinte veces, y ahora está escrito una.

**Las tres anomalías de la clase 1 vivían ahí**, en la diferencia entre veinte y uno.

# 6. Las filas que la base rechaza

Apunte, sección 5.6.

Pandas no hace cumplir ninguna regla. Así que acá escribimos a mano las tres revisiones que un DBMS
hace solo, para ver qué habría rechazado.

```python
def revisar_clave_primaria(tabla, columnas, nombre):
    dup = tabla.duplicated(subset=columnas, keep=False)
    nulos = tabla[columnas].isna().any(axis=1)
    if dup.any():
        return f"RECHAZADO en {nombre}: clave repetida en {dup.sum()} filas"
    if nulos.any():
        return f"RECHAZADO en {nombre}: clave con nulos en {nulos.sum()} filas"
    return f"ok {nombre}: clave primaria válida"

def revisar_clave_foranea(hija, col_hija, padre, col_padre, nombre):
    huerfanas = ~hija[col_hija].isin(padre[col_padre])
    if huerfanas.any():
        malos = sorted(int(v) for v in hija.loc[huerfanas, col_hija].unique())
        return f"RECHAZADO en {nombre}: {huerfanas.sum()} fila(s) apuntan a {malos}"
    return f"ok {nombre}: integridad referencial válida"

def revisar_dominio(tabla, columna, prueba, nombre):
    malos = tabla[columna].notna() & ~tabla[columna].apply(prueba)
    if malos.any():
        return f"RECHAZADO en {nombre}: {malos.sum()} valor(es) fuera del dominio"
    return f"ok {nombre}: dominio respetado"


print("tres revisiones definidas")
```

Tres funciones, una por integridad: de entidad, referencial y de dominio. Son las tres de la
sección 5.4 del apunte.

Sobre las seis tablas bien armadas, las tres tienen que pasar.

```python
print(revisar_clave_primaria(visita, ["hogar_id", "fecha"], "visita"))
print(revisar_clave_foranea(visita, "hogar_id", hogar, "id", "visita -> hogar"))
print(revisar_dominio(visita, "personas", lambda v: 1 <= v <= 20, "visita.personas"))
```

Las tres pasan. Ahora las tres cargas que un DBMS habría rechazado.

```python
# Caso 1: un segundo hogar 41, el del T8001.
print("CASO 1 — el hogar 41 duplicado del T8001")
hogar_41 = pd.DataFrame([{"id": 41, "localidad_id": 2}, {"id": 41, "localidad_id": 2}])
con_41 = pd.concat([hogar, hogar_41], ignore_index=True)
print("  ", revisar_clave_primaria(con_41, ["id"], "hogar"))
print()

# Caso 2: una visita a un hogar que no existe.
print("CASO 2 — una visita al hogar 99, que no existe")
visita_huerfana = pd.concat([visita, pd.DataFrame([{
    "hogar_id": 99, "fecha": "2025-07-01", "encuestador_legajo": "E01",
    "personas": 3, "ingreso": 190000, "cobertura_salud": "si",
    "satisfaccion": "buena"}])], ignore_index=True)
print("  ", revisar_clave_foranea(visita_huerfana, "hogar_id", hogar, "id", "visita -> hogar"))
print()

# Caso 3: personas = -3.
print("CASO 3 — personas = -3")
visita_mala = visita.copy()
visita_mala.loc[0, "personas"] = -3
print("  ", revisar_dominio(visita_mala, "personas", lambda v: 1 <= v <= 20, "visita.personas"))
```

Los tres rechazados, y cada mensaje dice **qué regla se violó y en cuántas filas**.

Ahora comparalo con lo que hizo el T8001 con los mismos tres defectos:

| Defecto | Con un archivo | Con las reglas declaradas |
|---|---|---|
| Dos filas del hogar 41 | entraron las dos; se descartaron después y se perdió un hogar | la segunda no entra |
| Una referencia a algo inexistente | no se puede ni detectar: no hay otra tabla | se rechaza en la carga |
| `personas = -3` | entra y corre el promedio | se rechaza en la carga |

Es el mismo dato y el mismo defecto. **Lo que cambia es el momento en que el error se descubre**, y
de eso depende si se puede arreglar: en la carga hay alguien con el formulario en la mano.

Y acordate del paso 2 del apunte: cuando una regla rechaza un dato, hay que decidir **quién está
mal**. Si existiera un hogar de 25 personas, el que está mal es el dominio y no el dato.
